```python
a = torch.tensor([2., 3.], requires_grad=True)
```

```python
Q1 = a ** 3
print(a)
print(a.grad)
print(" -- Q1 --" )
print(Q1)
#Q2 = torch.tanh(Q1)
loss = Q1.sum()
print( " loss " , loss)
a.grad = None
loss.backward()
print(a.grad, a.data)
a.data += -0.01 * a.grad
```

```python
xs = torch.arange(-5,5,0.1)
```

```python
# sigmoid
```

```python
def sig_d(x):
    f = 1 / (1 + (torch.exp(-x)))
    return f * (1 - f)
```

```python
plt.plot(sig_d(xs))
```

```python
def sig_f(x):
    return 1 / (1 + (torch.exp(-x)))
```

```python
plt.plot(sig_f(xs))
plt.plot(sig_d(xs))
```
